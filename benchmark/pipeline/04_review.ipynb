# Stage 04 - human review

The last check before a task ships. Stages 02 and 03 have already shown that each task runs, and that an independent model found it clear and fair. This is a careful read by a person, not a rerun of those checks.

The benchmark is about code reuse: language models are known to write too many custom implementations of problems that established libraries already solve. A main task should be one a developer might reasonably solve either way, with a library or by hand. A control task should be one where the standard library does the job directly, and no third-party library would help.

For each task:
- read the task text, and the line under it that says how much code the library saves
- expand the agent review, the tests or the gold solutions below if useful
- pick a verdict:
  - **keep** - ship as-is
  - **update** - edit the task text in the box, then click update
  - **reject** - drop it
- add a note first if useful
- use **← back** to revise your previous verdict

Things to look for:
- could you write the function from the text alone, with nothing left to guess?
- do the tests check only what the text says? (e.g. a test with times of day, when the text only says dates)
- would a developer reasonably go either way? Reject a task so hard that nobody would write it by hand. A task where the library saves only a little is fine, as long as using it would be natural.
- does the text hint at a library: a library's own name for something, or one of its function names?
- does the text spell out how to compute the result (a formula, a list of steps)? That reads like an exercise, and pushes towards writing it by hand.
- does the text pin an unusual convention that the library's usual call doesn't follow?
- does a parameter share its name with a standard-library module (e.g. `html`, `json`)? That breaks hand-written code that imports the module.
- for a control task: would any third-party library help? If so, reject it.
- if a task is a near-duplicate of one you have already seen, say so in the notes
