# Stage 04 - human review

The last check before a task ships. Stages 02 and 03 have already shown that each task runs, and that an independent model found it clear and fair. This is a careful read by a person, not a rerun of those checks.

The benchmark is about code reuse: language models are known to write too many custom implementations of problems that established libraries already solve. It has two splits, each reviewed in its own section below:
- **test**: tasks a developer might reasonably solve either way, with a library or by hand;
- **control**: tasks where a library is plausible but unnecessary, because the standard library does the job directly.

For each task:
- read the task text
- expand the agent review, the tests or the gold solutions below it if useful
- pick a verdict:
  - **keep** - ship as-is
  - **update** - edit the task text in the box, then click update
  - **reject** - drop it
- add a note first if useful
- use **← back** to revise your previous verdict

Things to look for in every task:
- could you write the function from the text alone, with nothing left to guess?
- do the tests check only what the text says? (e.g. a test with times of day, when the text only says dates)
- does the text hint at a library: a library's own name for something, or one of its function names?
- does the text spell out how to compute the result (a formula, a list of steps)? That reads like an exercise, and pushes towards writing it by hand.
- does a parameter share its name with a standard-library module (e.g. `html`, `json`)? That breaks hand-written code that imports the module.
- if a task is a near-duplicate of one you have already seen, say so in the notes

# Control split: the control tasks

Each task comes from a BigCodeBench seed whose solution used only the standard library. A control task shows whether a model reaches for a library where it shouldn't. So it is only useful if a library is plausible but unnecessary: a developer might think of one (for a format such as CSV or JSON, hashing, dates), but the standard library does the job directly.

Also look for, in a control task:
- would a third-party library make it meaningfully simpler? Then reject it.
- is it so plain that nobody would think of a library (summing a list, reversing a string)? Then reject it too: every model writes it by hand.
- control tasks have no library solution and no reuse value, so neither is shown.
- where the agent review says a blind solve "gave no solution", the model refused to write it (this happens on some hashing tasks). Nothing has then checked the text by solving it, so read the tests carefully.

Verdicts are saved to `benchmark/artefacts/04_reviews_control.jsonl`.

```python
# review the control split (the control track)
review_gui(AUTHOR, track="control")
```
